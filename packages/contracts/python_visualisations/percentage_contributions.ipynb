```python
import json


def analyze_contributions(json_file_path, fields):
    """
    Calculate percentage contributions of selected fields relative to their total sum for each entry.

    Args:
        json_file_path (str): Path to the JSON file.
        fields (list of str): Fields to analyze, e.g., ["SIGN_UP", "PROCESS_MESSAGES"].

    Returns:
        dict: For each field, a dict with min and max percentage contributions across all entries.
    """
    with open(json_file_path, "r") as f:
        data = json.load(f)

    field_percentages = {field: [] for field in fields}

    for entry_key, entry_value in data.items():
        values = [entry_value.get(field, 0) for field in fields]
        total = sum(values)

        if total == 0:
            continue  # Avoid division by zero

        if (entry_key == "1600_users_0_invalid_votes"):
            for field in fields:
                value = entry_value.get(field, 0)
                percent = (value / total) * 100
                field_percentages[field].append(percent)

    summary = {}
    for field, percentages in field_percentages.items():
        if percentages:
            summary[field] = {
                "min_percent": min(percentages),
                "max_percent": max(percentages)
            }
        else:
            summary[field] = {
                "min_percent": None,
                "max_percent": None
            }

    return summary
```

```python
FIELDS = [
    "SIGN_UP",
    "UPDATE_POLL",
    "JOIN_POLL",
    "TOTAL_VALID_VOTES",
    "PROCESS_MESSAGES",
    "TALLY_PROOFS"
]
analyze_contributions("custom_benchmark_data_start_at_400.json", FIELDS)
```
