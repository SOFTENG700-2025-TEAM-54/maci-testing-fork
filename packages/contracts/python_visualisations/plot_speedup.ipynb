```python
import json
import re
import matplotlib.pyplot as plt
from collections import defaultdict

# Extract and group timings by number of machines
machine_timings = defaultdict(list)
for key, value in data.items():
    match = re.search(r'MACHINES_(\d+)_', key)
    if match:
        machines = int(match.group(1))
        machine_timings[machines].append(value)

# Compute average total time for each machine count
average_times = {m: sum(times) / len(times) for m, times in machine_timings.items()}

# Calculate speedup relative to the single-machine baseline
base_time = average_times[1]
speedups = {m: base_time / t for m, t in average_times.items()}

# Prepare data for plotting
machines_sorted = sorted(speedups)
measured_speedup = [speedups[m] for m in machines_sorted]
ideal_speedup = machines_sorted  # ideal is linear speedup: speedup = num_machines

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(machines_sorted, measured_speedup, marker='o', label='Measured Speedup')
plt.plot(machines_sorted, ideal_speedup, linestyle='--', color='gray', label='Ideal Speedup')

plt.xlabel('Number of Machines')
plt.ylabel('Speedup')
plt.title('Measured vs Ideal Speedup')
plt.grid(True)
plt.legend()
plt.xticks(machines_sorted)
plt.tight_layout()
plt.show()
```
